## 1. Recipe Audit Station CSV

### Problem

Baca satu file CSV stasiun dan identifikasi masalah pada skema kolom,
rentang waktu, nilai hilang, serta timestamp duplikat. Hasilnya berupa
ringkasan QA/QC, bukan analisis iklim atau prediksi hujan. File sumber
tidak boleh diubah.

### Minimal Theory

Audit data dilakukan sebelum agregasi atau analisis agar kesalahan
struktur, waktu, dan nilai dapat diketahui lebih awal. Timestamp ganda
dapat memengaruhi agregasi, sedangkan nilai hilang atau tipe kolom yang
tidak sesuai dapat mengubah hasil analisis.

### Input dan Constraints

-   `CSV_PATH`: path file CSV.
-   `TIME_COLUMN`: nama kolom waktu.
-   `UNITS`: kontrak satuan untuk kolom yang relevan.
-   Pertahankan file sumber; gunakan pembacaan *read-only*.
-   Jangan menghapus atau memperbaiki baris sumber secara otomatis.
-   Jangan menyimpulkan hasil ilmiah yang tidak diuji oleh audit.

### Prompt dan Procedure

> Buat draft recipe untuk audit QA/QC station CSV. Periksa skema kolom,
> rentang waktu, nilai hilang, timestamp duplikat, dan tipe data tanpa
> mengubah input mentah. Nyatakan input, teori minimum, asumsi,
> prosedur, output yang diharapkan, rencana implementasi Python,
> pemeriksaan kegagalan, serta status eksekusi. Pisahkan spesifikasi
> recipe dari klaim bahwa program telah dijalankan.

Langkah kerja: 1. Baca CSV secara *read-only*. 2. Periksa apakah kolom
wajib tersedia. 3. Validasi tipe data dan rentang timestamp. 4. Hitung
nilai hilang dan timestamp duplikat. 5. Buat laporan QA/QC dan catatan
pemeriksaan. 6. Catat error atau kolom yang tidak sesuai tanpa mengubah
sumber.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan `pandas` untuk membaca CSV
dan menghitung ringkasan. Kode disimpan pada path relatif, misalnya
`code/recipes/io/r-io-001.py`. Path ini merupakan rancangan, bukan bukti
bahwa file sudah dibuat.

### Output

-   `results/io/station-summary.md`: laporan Markdown.
-   `results/run-records/r-io-001.json`: run record jika program
    dijalankan.

Laporan mencakup jumlah baris dan kolom, kolom yang ditemukan, rentang
waktu, jumlah nilai hilang, jumlah timestamp duplikat, dan status setiap
pemeriksaan.

### Optional Check / Evidence

Periksa bahwa file input tidak berubah, laporan tidak kosong, jumlah
baris sesuai dengan sumber, dan status eksekusi dicatat. Gunakan
`Not yet executed` jika program belum dijalankan. Label `Executed` hanya
digunakan jika bukti eksekusi tersedia.

## 3. Recipe Model Difusi Panas Satu Dimensi

### Problem

Buat recipe untuk mensimulasikan perubahan suhu terhadap posisi dan
waktu berdasarkan profil awal, difusivitas termal, kondisi batas, jarak
grid, dan langkah waktu.

### Minimal Theory

Persamaan difusi panas satu dimensi adalah:

\[
\frac{\partial T}{\partial t}=\alpha\frac{\partial^2T}{\partial x^2}
\]

dengan (T) adalah suhu, (t) waktu, (x) posisi, dan (\alpha)
difusivitas termal.

Untuk skema beda hingga eksplisit standar, parameter kestabilan dapat
ditulis:

\[ r=\frac{\alpha\Delta t}{(\Delta x)^2} \]

Pada kasus satu dimensi dengan skema eksplisit standar, syarat
kestabilan yang umum digunakan adalah (r \le 1/2). Syarat ini
perlu disesuaikan dengan skema numerik yang benar-benar digunakan.

### Input dan Constraints

-   `INITIAL_PROFILE`: profil suhu awal.
-   `ALPHA`: difusivitas termal dan satuannya.
-   `BOUNDARIES`: kondisi batas.
-   `DZ`: jarak grid.
-   `DT`: langkah waktu.
-   Pastikan konsistensi satuan.
-   Jangan mengklaim kalibrasi atau konvergensi tanpa pengujian.

### Prompt dan Procedure

> Buat draft recipe model difusi panas satu dimensi menggunakan profil
> awal `[INITIAL_PROFILE]`, difusivitas `[ALPHA]`, kondisi batas
> `[BOUNDARIES]`, jarak grid `[DZ]`, dan langkah waktu `[DT]`. Sertakan
> persamaan pengatur, konvensi satuan dan tanda, pemeriksaan kestabilan,
> pembanding analitik atau kasus dasar, profil output, grafik, run
> record, serta peringatan jika parameter atau dependency belum
> tersedia. Jangan menyatakan eksekusi tanpa bukti.

Langkah kerja: 1. Validasi parameter dan satuan. 2. Bentuk grid ruang
dan waktu. 3. Terapkan kondisi awal serta kondisi batas. 4. Periksa
kestabilan skema. 5. Jalankan iterasi jika implementasi tersedia. 6.
Bandingkan hasil dengan kasus dasar atau solusi analitik jika
memungkinkan. 7. Simpan profil suhu, grafik, dan catatan eksekusi.

### Code / Implementation Plan

Implementasi dapat menggunakan Python, `numpy` untuk komputasi numerik,
dan `matplotlib` untuk visualisasi. Pilihan metode numerik, parameter,
serta dependency harus dicatat.

### Output

-   Profil suhu terhadap posisi dan waktu.
-   Grafik hasil simulasi.
-   Ringkasan parameter.
-   Run record berisi status dan peringatan.

### Optional Check / Evidence

Periksa kestabilan numerik, kondisi batas, konsistensi satuan, dan
kesesuaian dengan kasus dasar. Jangan mengklaim konvergensi, kalibrasi,
atau keberhasilan eksekusi tanpa bukti yang sesuai.

# BAGIAN II --- RECIPE BARU BUATAN SAYA

## 4. Recipe QC dan Analisis Inversi Resistivitas Semu 1D (Geolistrik Sounding)

### Problem

Membuat workflow reproducible untuk memeriksa kualitas data kurva sounding resistivitas semu 1D (konfigurasi Schlumberger), memvisualisasikan hubungan antara separasi elektroda arus (AB/2) dengan resistivitas semu (rho_a), serta melakukan pemodelan maju (forward modeling) atau inversi sederhana untuk mengestimasi ketebalan dan resistivitas sebenarnya dari lapisan bawah permukaan.

Masalah yang diperiksa meliputi nilai hilang, resistivitas semu bernilai negatif atau nol, data AB/2 yang tidak berurutan, duplikasi bentangan elektroda, serta outlier akustik/spurious akibat gangguan kabel/kontak elektroda. Tujuannya adalah memastikan data layak diinversi sebelum dilakukan interpretasi hidrogeologi atau geologi teknik.

### Minimal Theory

Dalam pengukuran Geolistrik konfigurasi Schlumberger, resistivitas semu (\rho_a) dihitung menggunakan faktor geometri (K):

\[ \rho_a = K \cdot \frac{\Delta V}{I} \]

dengan faktor geometri K untuk konfigurasi Schlumberger ditentukan oleh setengah jarak elektroda arus (s = AB/2) dan setengah jarak elektroda potensiometer (b = MN/2):

\[ K = \frac{\pi (s^2 - b^2)}{2b} \]

Hubungan antara resistivitas semu terukur terhadap spasial AB/2 digambarkan dalam skala log-log (kurva sounding). Pemodelan 1D diasumsikan terdiri dari N lapisan homogen isotropik dengan resistivitas \rho_i dan ketebalan h_i.

Tingkat kesalahan pemodelan (root mean square error / RMSE) antara resistivitas semu observasi (\rho_{a, obs}) dan kalkulasi (\rho_{a, cal}) dihitung sebagai:

\[ \text{RMSE} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} \left( \frac{\rho_{a, obs}^{(i)} - \rho_{a, cal}^{(i)}}{\rho_{a, obs}^{(i)}} \right)^2} \times 100\% \]

Resistivitas semu bersifat tak tunggal (non-uniqueness / ekivalensi). Nilai resistivitas tanah tidak secara langsung menentukan jenis batuan atau akuifer tanpa didukung informasi sumur bor atau geologi lokal.

### Input dan Constraints

Input utama:
- `DATA_PATH`: lokasi file CSV data geolistrik.
- `POINT_ID`: nama titik sounding/dugaan (misal: VES-01).
- `AB2`: setengah jarak elektroda arus (m).
- `MN2`: setengah jarak elektroda tegangan (m).
- `R_RHO`: resistivitas semu terukur (ohm.m) atau nilai resistansi R = \Delta V / I.

Input tambahan jika tersedia:
- `COORDINATES`: koordinat titik sounding (UTM/Geografis).
- `ELEVATION`: elevasi muka tanah (m).
- `DRILL_LOG`: data log bor terdekat sebagai kontrol batas lapisan.

Batasan:
1. Berkas mentah harus bersifat read-only.
2. Nilai \rho_a \le 0 dianggap tidak valid dan harus ditandai.
3. Urutan data harus disortir berdasarkan nilai AB/2 yang meningkat secara fisis.
4. Nilai MN/2 harus selalu lebih kecil dari AB/2 (MN/2 < AB/2).
5. Grafik sounding wajib digambarkan pada skala logaritma ganda (log-log).
6. Hasil pemodelan Inversi 1D harus menyertakan persentase RMS Error.
7. Dilarang memastikan jenis akuifer atau batas batuan tanpa konfirmasi geologi regional.

### Prompt dan Procedure

**Prompt yang dapat digunakan kembali:**

> Anda bertindak sebagai asisten pengolahan data geofisika eksplorasi geolistrik. Baca berkas `[DATA_PATH]` secara read-only. Periksa kelengkapan kolom wajib (`AB2`, `MN2`, `R_RHO`), tipe data, nilai tak-fisis (resistivitas negatif atau nol), duplikasi geometri, serta kondisi MN/2 \ge AB/2.
>
> Buat ringkasan laporan QA/QC data sebelum inversi. Tampilkan kurva sounding resistivitas semu observasi terhadap AB/2 pada skala log-log. Jalankan algoritma pemodelan majuan/inversi 1D sederhana untuk mengestimasi ketebalan (h) dan resistivitas (\rho) tiap lapisan bawah permukaan.
>
> Sertakan kurva kalkulasi hasil fitting, hitung RMS Error, dan tampilkan profil penampang 1D (staircase plot). Pisahkan secara tegas antara data mentah terukur, hasil estimasi numerik, dan interpretasi hidrogeologi. Buat run record jika kode dieksekusi, atau tulis Not yet executed jika program belum dijalankan.

**Procedure:**
1. Baca data geolistrik mentah secara read-only.
2. Lakukan audit skema: validasi kolom, periksa nilai negatif/nol, dan cek rasio MN/2 < AB/2.
3. Bersihkan atau eliminasi titik ekstrim (spurious noise) serta urutkan berdasarkan bentangan AB/2.
4. Plot kurva sounding observasi log-log (Resistivitas Semu vs AB/2).
5. Tentukan model tebakan awal (initial model) jumlah lapisan, ketebalan, dan resistivitas.
6. Jalankan pemodelan maju/inversi 1D (menggunakan transform Digital Filter Linear Filter Ghosh/Guptasarma).
7. Hitung nilai persentase kecocokan (RMS Error) antara data observasi dan model.
8. Plot kurva fitting log-log beserta penampang tebal-resistivitas 1D (staircase plot).
9. Susun laporan QC, tabel parameter lapisan hasil inversi, dan catat status eksekusi.

### Code / Implementation Plan

Rencana implementasi menggunakan Python:
- `pandas`: ekstraksi data tabular dan pemeriksaan QA/QC.
- `numpy`: perhitung numerik filter konvolusi linear geolistrik.
- `matplotlib`: plot kurva log-log sounding dan staircase plot 1D.
- `scipy.optimize`: fungsi minimisasi non-linear (misal `curve_fit` atau `least_squares`) untuk proses inversi.
- `json` & `pathlib`: pengelolaan metadata dan penyimpan catatan eksekusi.

Rancangan struktur direktori proyek:

``` text
geoelectric-sounding-qc/
├── data/
│   └── ves-sounding-data.csv
├── code/
│   └── ves_inversion_qc.py
└── results/
    ├── ves-qc-report.md
    ├── sounding-curve-fit.png
    ├── inverted-layers.csv
    └── run-record.json
```

Struktur di atas merupakan acuan perancangan. Keberadaan berkas fisik harus diverifikasi secara terpisah setelah eksekusi dilakukan.

### Output

1. `ves-qc-report.md`: laporan QA/QC berisi statistik data mentah, jumlah titik valid, peringatan geometri, dan nilai RMSE inversi.
2. `sounding-curve-fit.png`: grafik log-log perbandingan kurva sounding terukur vs kurva kalkulasi model beserta penampang 1D resistivitas terhadap kedalaman.
3. `inverted-layers.csv`: tabel estimasi resistivitas (Ohm.m), ketebalan (m), dan kedalaman (m) per lapisan.
4. `run-record.json`: catatan eksekusi berisi waktu run, konfigurasi parameter inversi, RMS error, dan log sistem.

### Optional Check / Evidence

Checklist pemeriksaan:
- [ ] File CSV input tidak diubah (read-only).
- [ ] Nilai AB/2 dan MN/2 diperiksa rasio geometrinya.
- [ ] Data resistivitas semu bernilai non-positif berhasil terdeteksi.
- [ ] Grafik disajikan dalam domain log-log dengan satuan Ohm.m dan meter.
- [ ] Perhitungan RMSE inversi disajikan secara objektif.
- [ ] Ambiguitas ekivalensi resistivitas dinyatakan dalam laporan.
- [ ] Status eksekusi ditandai secara akurat (Draft / Not yet executed / Executed).

Status eksekusi yang diperbolehkan:
- `Draft`: spesifikasi recipe sedang dirancang.
- `Not yet executed`: skrip/program belum dijalankan.
- `Partial`: pemeriksaan QC selesai, namun tahap inversi belum berjalan.
- `Executed`: skrip telah dieksekusi penuh dan berkas output tersedia.

### Keterbatasan dan Risiko

Hasil inversi 1D geolistrik sangat rentan terhadap prinsip ekivalensi (equivalence) dan supresi (suppression), di mana kombinasi tebal dan resistivitas yang berbeda dapat menghasilkan kurva respon yang hampir serupa. Asumsi lapisan datar homogen lateral (1D) sering kali terlanggar pada struktur bawah permukaan yang kompleks/miring. Estimasi lapisan dari inversi 1D harus dianggap sebagai model matematis awal dan memerlukan verifikasi log pemboran atau metode geofisika pelengkap (seperti Seismik atau TDEM).

### Ringkasan Recipe Baru

Recipe ini menyusun kerangka kerja standar dalam pengolahan data geolistrik sounding 1D Schlumberger. Dimulai dari validasi kualitas data mentah dan faktor geometri, visualisasi logaritmis kurva sounding, hingga pemodelan inversi kuantitatif dengan kontrol RMS Error yang jelas. Recipe ini memastikan proses pengolahan data transparan, terukur, dan terhindar dari interpretasi yang prematur.
